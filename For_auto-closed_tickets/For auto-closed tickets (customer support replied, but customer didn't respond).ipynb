# Importing necessary libraries

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as  plt
import seaborn as sns
import plotly.express as px
from plotly.offline import init_notebook_mode, iplot
init_notebook_mode(connected=True)

```

# Loading Dataset

```python
df = pd.read_csv("auto_close_tickets.csv")
```

# Exploratory Data Analysis(EDA)

```python
df.head()
```

```python
df.info()
```

```python
sunburst_data = df.groupby(['Age Group', 'Ticket Type', 'Ticket Subject']).size().reset_index(name='Count')

fig = px.sunburst(
    sunburst_data,
    path=['Age Group', 'Ticket Type', 'Ticket Subject'],
    values='Count',
    title="Sunburst Chart: Ticket Distribution by Age Group, Type, and Subject"
)

fig.update_layout(
    paper_bgcolor="#E6E6FA",      
          
)
iplot(fig)

# Note: Run all previous cells first, then this one. Hover over the sunburst plot to explore detailed insights.
```

```python
# Counting the number of each ticket type
ticket_counts = df['Ticket Type'].value_counts()

# Plotting horizontal bar chart
plt.figure(figsize=(6,4),facecolor= 'yellow')
ax= ticket_counts.plot(kind='barh',color="brown")
ax.set_facecolor("#1CC4AF")

# plot
plt.title('Ticket Type Distribution')
plt.xlabel('Number of Tickets')
plt.ylabel('Ticket Type')
plt.tight_layout()
plt.show()
```

  <h2 style="color:#800000;">90-Day Auto-Close Rule for Tickets</h2>

  <p><b>Why we needed this rule:</b><br>
  Sometimes, customers send a ticket and our customer support team replies, but the customer does not respond back. These tickets remain open even though no one is following up. This causes clutter in the system and makes it hard to focus on the tickets that really need attention.</p>

  <p><b>What we did:</b><br>
  We created an automated rule: If the customer support team has already responded and the customer does not reply within <b>90 days</b>, then the system will automatically mark the ticket as resolved and close it.</p>

  <p><b>How this helps:</b></p>
  <ul style="color:#2F4F4F;">
    <li>It keeps the support system clean by closing inactive tickets.</li>
    <li>Support agents can focus more on real, open, and urgent issues.</li>
    <li>It saves time and improves the speed of resolving other tickets.</li>
    <li>If the customer still needs help, they can easily raise a new ticket at any time.</li>
  </ul>

  <p><b>Overall Benefit:</b><br>
  This rule improves the efficiency and organization of customer service, so we can serve customers better and faster.</p>

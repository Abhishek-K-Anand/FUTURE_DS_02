# Importing necessary libraries

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as  plt
import seaborn as sns
import plotly.express as px
from plotly.offline import init_notebook_mode, iplot
init_notebook_mode(connected=True)
```

# Loading dataset

```python
df = pd.read_csv("open_tickets.csv")
```

# Exploratory Data Analysis(EDA)

```python
df.head()
```

```python
df.info()
```

```python
sunburst_data = df.groupby(['Age Group', 'Ticket Type', 'Ticket Subject']).size().reset_index(name='Count')

fig = px.sunburst(
    sunburst_data,
    path=['Age Group', 'Ticket Type', 'Ticket Subject'],
    values='Count',
    title="Sunburst Chart: Ticket Distribution by Age Group, Type, and Subject"
)

fig.update_layout(
    paper_bgcolor="#E6E6FA",      
          
)
iplot(fig)

# Note: Run all previous cells first, then this one. Hover over the sunburst plot to explore detailed insights.
```

```python
# Counting the number of each ticket type
ticket_counts = df['Ticket Type'].value_counts()

# Plotting horizontal bar chart
plt.figure(figsize=(6,4),facecolor= 'yellow')
ax= ticket_counts.plot(kind='barh',color="#F0F0F0")
ax.set_facecolor("#1CC4AF")

# plot
plt.title('Ticket Type Distribution')
plt.xlabel('Number of Tickets')
plt.ylabel('Ticket Type')
plt.tight_layout()
plt.show()
```

  <h2 style="color:#997a00;">Summary of Unresolved Ticket Data (No Initial Response)</h2>

  <p><b>What we found:</b><br>
  In the <b>open_tickets.csv</b> file, several tickets were found with <b>no initial response</b> from the support team. These tickets remain open and unattended, which can lead to poor customer experience and dissatisfaction.</p>

  <p><b>Why this matters:</b><br>
  When customers don’t receive a response at all, it creates a negative impression of the support process. Delayed or absent replies reduce trust and may result in loss of customer loyalty.</p>

  <p><b>What we can do:</b></p>
  <ul style="color:#665500;">
    <li><b>Set up an alert system</b> to flag all tickets that remain unanswered for more than a few hours or a day, so they can be escalated quickly.</li>
    <li><b>Assign higher priority</b> to such tickets and regularly review them in team meetings to ensure no ticket goes unnoticed.</li>
    <li><b>Train support staff</b> to respond promptly, even if only to acknowledge the ticket and let the customer know their issue is being reviewed.</li>
    <li><b>Use a chatbot</b> to send immediate auto-replies or collect basic information from the customer while they wait. This builds confidence and keeps them engaged.</li>
    <li><b>Monitor response times</b> and introduce a KPI (Key Performance Indicator) for “First Response Time” to drive accountability.</li>
    <li><b>Review ticket assignment logic</b> to ensure open tickets aren’t left unassigned or stuck in the system.</li>
  </ul>

  <p><b>Overall Impact:</b><br>
  By focusing on unresolved tickets and taking proactive steps to respond quickly, we can significantly improve customer satisfaction, prevent escalation of issues, and maintain a positive brand image.</p>
